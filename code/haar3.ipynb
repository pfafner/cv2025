# Image (OpenCV)

```python
img = cv2.imread(imagePath)
img.shape
```

```python
img = cv2.resize(img, (img.shape[1]//2, img.shape[0]//2))
img.shape
```

```python
gray_image = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
gray_image.shape
```

# Face Cascade Classifier

```python
clock_classifier  = cv2.CascadeClassifier("haarcascade_wallclock.xml")
```

```python
clocks = clock_classifier.detectMultiScale(gray_image, scaleFactor=1.1, minNeighbors=1, minSize=(100, 100))
```

```python
clocks.shape
```

```python
clocks
```

```python
for (x, y, w, h) in clocks:
    cv2.rectangle(img, (x, y), (x + w, y + h), (0, 255, 0), 2)
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
plt.figure(figsize=(20,10))
plt.imshow(img_rgb)
plt.axis('off')
```
