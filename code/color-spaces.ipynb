```python
import numpy as np
import matplotlib.pyplot as plt
import skimage as ski
```

```python
from skimage.color import rgb2hsv, hsv2rgb
```

```python
I = plt.imread('quetzal.png')
I.shape
```

```python
# Convert RGB 2 grayscale
def rgb2gray_mean(I):
    ''' Y = 0.3333 R + 0.3333 G + 0.3333 B. '''
    (h, w, c) = I.shape
    if (c > 3):
        I = I[:,:,3]
    
    w = np.ones(3)/3.
    Igray = I @ w
    return Igray
```

```python
# Convert RGB 2 grayscale
def rgb2gray(I):
    ''' Y = 0.2125 R + 0.7154 G + 0.0721 B. '''
    (h, w, c) = I.shape
    if (c > 3):
        I = I[:,:,3]
    
    w = np.array([0.02125, 0.7154, 0.0721])
    Igray = I @ w
    return Igray
```

```python
Igray = rgb2gray(I)
```

```python
plt.figure(figsize=(10,5))
plt.subplot(1,3,1)
plt.imshow(rgb2gray_mean(I), cmap='gray')
plt.subplot(1,3,2)
plt.imshow(I)
plt.subplot(1,3,3)
plt.imshow(rgb2gray(I), cmap='gray')
plt.show()
```
