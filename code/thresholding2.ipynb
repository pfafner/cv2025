```python
import numpy as np
import matplotlib.pyplot as plt
import skimage as ski
```

```python
from skimage.filters.thresholding import threshold_isodata, threshold_li, threshold_local, threshold_mean, threshold_minimum, threshold_multiotsu, threshold_niblack, threshold_otsu, threshold_sauvola, threshold_triangle
```

## Thresholding methods

```python
#I = plt.imread('rice.jpg')
I = plt.imread('document.png')
I.shape
```

```python
I = I[:,:,:3]
```

```python
I = ski.color.rgb2gray(I)
I.shape
```

```python
print(I.min(), I.max())
```

```python
I = (I * 255).astype(np.uint8)
```

```python
h = np.histogram(I, bins=256)
```

```python
pixnum = I.shape[0] * I.shape[1]
pixnum
```

```python
plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.imshow(I, cmap='gray', vmin=0, vmax=255)
plt.subplot(1,3,2)
plt.title('Histogram')
plt.plot(h[1][1:], h[0])
plt.plot(h[1][1:], h[0].cumsum())
plt.subplot(1,3,3)
plt.title('Normalized Histogram')
#plt.plot(h[1][1:], h[0] / pixnum)
plt.plot(h[1][1:], h[0] / h[0].max())
plt.plot(h[1][1:], h[0].cumsum() / pixnum)
plt.show()
```

```python
th = 140
NewI = I.copy()
NewI[I < th] = 0
NewI[I >= th] = 255
```

```python
NewI.shape
```

```python
plt.figure(figsize=(9,4))
plt.subplot(1,2,1)
plt.title('Original Image')
plt.imshow(I, cmap='gray')
plt.subplot(1,2,2)
plt.title('Binarized Image')
plt.imshow(NewI, cmap='gray')
plt.show()
```

## Otsu

```python
th = threshold_otsu(I)
Ith = 255 * (I >= th).astype(np.uint8)
print(th)
```

```python
plt.figure(figsize=(9,4))
plt.subplot(1,2,1)
plt.imshow(I, cmap='gray', vmin=0, vmax=255)
plt.subplot(1,2,2)
plt.imshow(Ith, cmap='gray', vmin=0, vmax=255)
plt.show()
```
