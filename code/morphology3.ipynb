```python
import numpy as np
import matplotlib.pyplot as plt

import skimage as ski
from skimage import morphology as mo
```

## Morphological transformations

```python
def cross(r=2):
    x = np.zeros((r,r)).astype(int)
    x[:,r//2] = 1
    x[r//2,:] = 1
    x = np.pad(x, [1,1])
    return x
```

```python
r = 3
sq = mo.square(r)
st = mo.star(r)
dm = mo.diamond(r)
di = mo.disk(r)
oc = mo.octagon(r, 2)
cr = cross(r)
```

```python
plt.figure()
plt.subplot(2,3,1)
plt.imshow(np.pad(sq, [1,1]), cmap='gray')
plt.subplot(2,3,2)
plt.imshow(di, cmap='gray')
plt.subplot(2,3,3)
plt.imshow(st, cmap='gray')
plt.subplot(2,3,4)
plt.imshow(dm, cmap='gray')
plt.subplot(2,3,5)
plt.imshow(oc, cmap='gray')
plt.subplot(2,3,6)
plt.imshow(cr, cmap='gray')
plt.show()
```

## Dilation

```python
I = plt.imread('quetzalgris.png')
I.shape
```

```python
I = ski.color.rgb2gray(I[:,:,:3])
I.shape
```

```python
se3  = mo.disk(3)
se5  = mo.disk(5)
se7  = mo.disk(7)
se11 = mo.disk(11)
```

```python
dilation3  = mo.dilation(I, se3)
dilation5  = mo.dilation(I, se5)
dilation7  = mo.dilation(I, se7)
dilation11 = mo.dilation(I, se11)
```

```python
plt.figure(figsize=(10,4))
plt.subplot(1,4,1)
plt.imshow(I, cmap='gray')
plt.subplot(1,4,2)
plt.imshow(dilation3, cmap='gray')
plt.subplot(1,4,3)
plt.imshow(dilation5, cmap='gray')
plt.subplot(1,4,4)
plt.imshow(dilation7, cmap='gray')
plt.show()
```

## Morphological Gradient

```python
se3  = mo.disk(3)
se5  = mo.disk(5)
se7  = mo.disk(7)
se11 = mo.disk(11)
```

```python
gr3  = mo.dilation(I, se3) - mo.erosion(I, se3)
gr5  = mo.dilation(I, se5) - mo.erosion(I, se5)
gr7  = mo.dilation(I, se7) - mo.erosion(I, se7)
gr11 = mo.dilation(I, se11) - mo.erosion(I, se11)
```

```python
plt.figure(figsize=(10,4))
plt.subplot(1,4,1)
plt.imshow(I, cmap='gray')
plt.subplot(1,4,2)
plt.imshow(gr3, cmap='gray')
plt.subplot(1,4,3)
plt.imshow(gr5, cmap='gray')
plt.subplot(1,4,4)
plt.imshow(gr7, cmap='gray')
plt.show()
```
