```python
import numpy as np

import matplotlib.pyplot as plt
import skimage as ski
from PIL import Image

from skimage.color import rgb2hsv, hsv2rgb
```

# There is no red in this image

```python
I = plt.imread('fresas.png')
I = I[:,:,:3]
```

```python
I.shape
```

```python
print(I.min(), I.max())
```

```python
plt.figure(figsize=(5,5))
plt.imshow(I)
plt.show()
```

```python
# Convert from RGB to HSV
Ihsv = rgb2hsv(I)
```

```python
# plot HSV
plt.figure(figsize=(5,5))
plt.imshow(Ihsv)
plt.show()
```

```python
# Canales RGB y HSV

R = I[:,:,0]
G = I[:,:,1]
B = I[:,:,2]

H = Ihsv[:,:,0]
S = Ihsv[:,:,1]
V = Ihsv[:,:,2]
```

```python
print(H.min(), H.max())
print(S.min(), S.max())
print(V.min(), V.max())
```

```python
# reescale to [0,255] if scale is not [0,1]
maxx = I.max()
if (maxx > 1): maxx = 255
```

```python
plt.figure(figsize=(12,5))
plt.subplot(1,3,1)
plt.imshow(R, cmap='gray', vmin=0, vmax=maxx)
plt.title('R')
plt.subplot(1,3,2)
plt.imshow(G, cmap='gray', vmin=0, vmax=maxx)
plt.title('G')
plt.subplot(1,3,3)
plt.imshow(B, cmap='gray', vmin=0, vmax=maxx)
plt.title('B')
plt.show()
```

```python
plt.figure(figsize=(12,5))
plt.subplot(1,3,1)
plt.imshow(H, cmap='gray', vmin=0, vmax=1)
plt.title('H')
plt.subplot(1,3,2)
plt.imshow(S, cmap='gray', vmin=0, vmax=1)
plt.title('S')
plt.subplot(1,3,3)
plt.imshow(V, cmap='gray', vmin=0, vmax=1)
plt.title('V')
plt.show()
```

```python
# Histograms
r = np.histogram(R, bins=256)
g = np.histogram(G, bins=256)
b = np.histogram(B, bins=256)

h = np.histogram(H, bins=256)
s = np.histogram(S, bins=256)
v = np.histogram(V, bins=256)
```

```python
plt.figure()
plt.plot(r[1][:-1], r[0], 'r-', label='R')
plt.plot(g[1][:-1], g[0], 'g-', label='G')
plt.plot(b[1][:-1], b[0], 'b-', label='B')
plt.xlim([0, maxx])
plt.legend()
plt.show()
```

```python
plt.figure()
plt.plot(h[1][:-1], h[0], 'k-', label='hue')
plt.xlim([0,1])
plt.legend()
plt.show()
```
